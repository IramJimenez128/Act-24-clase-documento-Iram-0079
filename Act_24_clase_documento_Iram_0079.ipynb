```python
# Definimos la clase Documento
class Documento:
    def __init__(self, titulo, autor, fechaCreacion, formato):
        # Atributos del documento
        self.titulo = titulo
        self.autor = autor
        self.fechaCreacion = fechaCreacion
        self.formato = formato

    # Función para simular la edición
    def editar(self):
        nuevo_titulo = input(f"Introduce el nuevo título para '{self.titulo}': ")
        self.titulo = nuevo_titulo
        print("Sugerencia: El título ha sido modificado.")

    # Función para simular guardar
    def guardar(self):
        print(f"Archivo '{self.titulo}.{self.formato}' guardado exitosamente por {self.autor}.")

    # Función para imprimir los detalles
    def imprimir(self):
        print("\n--- Detalles del Documento ---")
        print(f"Título: {self.titulo}")
        print(f"Autor: {self.autor}")
        print(f"Fecha: {self.fechaCreacion}")
        print(f"Formato: {self.formato}")
        print("------------------------------\n")

# --- Programa Interactivo ---

print("Bienvenido al sistema de gestión de documentos\n")

# Creación del primer objeto (Interactivo)
print("Configurando el Primer Documento:")
doc1 = Documento(
    input("Título: "),
    input("Autor: "),
    input("Fecha (dd/mm/aaaa): "),
    input("Formato (pdf/docx): ")
)

# Creación del segundo objeto (Interactivo)
print("\nConfigurando el Segundo Documento:")
doc2 = Documento(
    input("Título: "),
    input("Autor: "),
    input("Fecha (dd/mm/aaaa): "),
    input("Formato (pdf/docx): ")
)

# Probando las funciones con el primer documento
print("\n--- Acciones sobre el Documento 1 ---")
doc1.imprimir()
doc1.editar()
doc1.guardar()

# Probando las funciones con el segundo documento
print("\n--- Acciones sobre el Documento 2 ---")
doc2.imprimir()
doc2.guardar()
```
